```python
import pandas as pd
import numpy as np
import re
from pathlib import Path
import matplotlib.pyplot as plt

COLS = ["frame_id", "addr", "retcode", "distances", "raw_mm", "filtered_mm", "filtered_in", "p0_str", "tnow", "height_level"]

def read_log(path):
    df = pd.read_csv(path, names=COLS, skiprows=1)
    df["source_file"] = str(path)
    df["raw_valid"] = df["raw_mm"].replace(-1, np.nan)
    df["raw_in"] = df["raw_valid"] / 25.4
    df["filtered_valid_in"] = df["filtered_in"].where(df["raw_mm"] != -1)
    df["time_s"] = (df["tnow"] - df["tnow"].iloc[0]) / 1000
    return df
```

```python
log_12_1 = "mobile_field_dataset_calibrated/bed1/mobile_12_1.txt"
log_12_2 = "mobile_field_dataset_calibrated/bed1/mobile_12_2.txt"
log_12_3 = "mobile_field_dataset_calibrated/bed1/mobile_12_3.txt"

df_1 = read_log(log_12_1)

df_2 = read_log(log_12_2)

df_3 = read_log(log_12_3)

# Plot raw & filtered vs. time

plt.figure(figsize=(12, 6))

plt.plot(df_1["time_s"], df_1["filtered_in"],
         label="Trial 1 (filtered)",
         color="tab:red",
         linewidth=2)

plt.plot(df_2["time_s"], df_2["filtered_in"],
         label="Trial 2 (filtered)",
         color="tab:green",
         linewidth=2)

plt.plot(df_3["time_s"], df_3["filtered_in"],
         label="Trial 3 (filtered)",
         color="tab:blue",
         linewidth=2)

plt.xlabel("Time (s)")
plt.ylabel("Filtered Distance (inches)")
plt.title("Overlaid 12 in. Runs")
plt.legend()
plt.grid(True)

plt.show()
```
