```python
import pandas as pd
import numpy as np
import re
from pathlib import Path
import matplotlib.pyplot as plt

COLS = ["frame_id", "addr", "retcode", "distances", "raw_mm", "filtered_mm", "filtered_in", "p0_str", "tnow", "height_level"]

def read_log(path):
    df = pd.read_csv(path, names=COLS, skiprows=1)
    df["source_file"] = str(path)
    df["raw_valid"] = df["raw_mm"].replace(-1, np.nan)
    df["raw_in"] = df["raw_valid"] / 25.4
    df["filtered_valid_in"] = df["filtered_in"].where(df["raw_mm"] != -1)
    return df
```

```python
log_file = "mobile_ground_dataset/mobile_16.txt"

df = read_log(log_file)

# Remove no peak data, convert to inches, align filtered values with valid raw values, and calculate time
df["raw_valid"] = df["raw_mm"].replace(-1, np.nan)
df["raw_in"] = df["raw_valid"] / 25.4
df["filtered_valid_in"] = df["filtered_in"].where(df["raw_mm"] != -1)
df["time_s"] = (df["tnow"] - df["tnow"].iloc[0]) / 1000

# Plot raw & filtered vs. time

plt.figure(figsize=(12, 6))

plt.plot(df["time_s"], df["raw_in"],
         label="Raw",
         color="tab:red",
         alpha=0.7)

plt.plot(df["time_s"], df["filtered_in"],
         label="Filtered",
         color="tab:green",
         linewidth=2)

plt.xlabel("Time (s)")
plt.ylabel("Distance (inches)")
plt.title("Raw vs Filtered Sensor Reading")
plt.legend()
plt.grid(True)

plt.show()
```
