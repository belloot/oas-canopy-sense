```python
import pandas as pd
import numpy as np
import re
from pathlib import Path
import matplotlib.pyplot as plt

COLS_MANUAL = ["manual_in"]
COLS_LIVE = ["frame_id", "addr", "retcode", "distances", "raw_mm", "filtered_mm", "filtered_in", "p0_str", "tnow", "height_level"]

def read_log_manual(path):
    df = pd.read_csv(path, names=COLS_MANUAL, skiprows=1)
    df["source_file"] = str(path)
    return df

def read_log_live(path):
    df = pd.read_csv(path, names=COLS_LIVE, skiprows=1)
    df["source_file"] = str(path)
    df["raw_valid"] = df["raw_mm"].replace(-1, np.nan)
    df["raw_in"] = df["raw_valid"] / 25.4
    df["filtered_valid_in"] = df["filtered_in"].where(df["raw_mm"] != -1)
    return df
```

```python
# For 12 inches
log_file_manual_12 = "manual_12_inches.txt"
log_file_live_12_1 = "mobile_field_dataset_calibrated/bed1/mobile_12_1.txt"
log_file_live_12_2 = "mobile_field_dataset_calibrated/bed1/mobile_12_2.txt"
log_file_live_12_3 = "mobile_field_dataset_calibrated/bed1/mobile_12_3.txt"

df_manual_12 = read_log_manual(log_file_manual_12)
df_live_12_1 = read_log_live(log_file_live_12_1)
df_live_12_2 = read_log_live(log_file_live_12_2)
df_live_12_3 = read_log_live(log_file_live_12_3)

manual_avg = df_manual_12["manual_in"].mean().round(2)
live_1_avg = df_live_12_1["filtered_valid_in"].mean().round(2)
live_2_avg = df_live_12_2["filtered_valid_in"].mean().round(2)
live_3_avg = df_live_12_3["filtered_valid_in"].mean().round(2)

print("Manual Measurements: " + str(len(df_manual_12["manual_in"])))
print("Mean Manual Average: " + str(manual_avg) + " inches")
print("12_1 Average: " + str(live_1_avg) + " inches")
print("12_2 Average: " + str(live_2_avg) + " inches")
print("12_3 Average: " + str(live_3_avg) + " inches")

print("12_1 vs. manual average: " + str((live_1_avg - manual_avg).round(2)) + " inches")
print("12_2 vs. manual average: " + str((live_2_avg - manual_avg).round(2)) + " inches")
print("12_3 vs. manual average: " + str((live_3_avg - manual_avg).round(2)) + " inches")
```

```python
# For 20 inches
log_file_manual_20 = "manual_20_inches.txt"
log_file_live_20_1 = "mobile_field_dataset_calibrated/bed1/mobile_20_1.txt"
log_file_live_20_2 = "mobile_field_dataset_calibrated/bed1/mobile_20_2.txt"
log_file_live_20_3 = "mobile_field_dataset_calibrated/bed1/mobile_20_3.txt"

df_manual_20 = read_log_manual(log_file_manual_20)
df_live_20_1 = read_log_live(log_file_live_20_1)
df_live_20_2 = read_log_live(log_file_live_20_2)
df_live_20_3 = read_log_live(log_file_live_20_3)

manual_avg = df_manual_20["manual_in"].mean().round(2)
live_1_avg = df_live_20_1["filtered_valid_in"].mean().round(2)
live_2_avg = df_live_20_2["filtered_valid_in"].mean().round(2)
live_3_avg = df_live_20_3["filtered_valid_in"].mean().round(2)

print("Manual Measurements: " + str(len(df_manual_20["manual_in"])))
print("Mean Manual Average: " + str(manual_avg) + " inches")
print("20_1 Average: " + str(live_1_avg) + " inches")
print("20_2 Average: " + str(live_2_avg) + " inches")
print("20_3 Average: " + str(live_3_avg) + " inches")

print("20_1 vs. manual average: " + str((live_1_avg - manual_avg).round(2)) + " inches")
print("20_2 vs. manual average: " + str((live_2_avg - manual_avg).round(2)) + " inches")
print("20_3 vs. manual average: " + str((live_3_avg - manual_avg).round(2)) + " inches")
```

```python
def summarize_file(df, truth_in):
    # If we're doing in-field runs, use the truth_manual
    # If we're doing ground control runs, use truth_in
    
    if truth_in == 12:
        truth_manual = 11.59
    elif truth_in == 16:
        truth_manual = 15.58
    elif truth_in == 20:
        truth_manual = 19.83

    raw_err = df["raw_in"] - truth_manual
    filt_err = df["filtered_valid_in"] - truth_manual

    return {
        "samples": len(df),
        "valid_samples": df["raw_valid"].notna().sum(),
        "invalid_samples": df["raw_valid"].isna().sum(),
        "invalid_pct": (100 * df["raw_valid"].isna().mean()).round(3),

        "raw_mean": df["raw_in"].mean().round(3),
        "raw_median": df["raw_in"].median().round(3),
        "raw_std": df["raw_in"].std().round(3),
        "raw_bias": raw_err.mean().round(3),
        "raw_mae": raw_err.abs().mean().round(3),
        "raw_rmse": np.sqrt((raw_err ** 2).mean()).round(3),

        "filtered_mean": df["filtered_valid_in"].mean().round(3),
        "filtered_median": df["filtered_valid_in"].median().round(3),
        "filtered_std": df["filtered_valid_in"].std().round(3),
        "filtered_bias": filt_err.mean().round(3),
        "filtered_mae": filt_err.abs().mean().round(3),
        "filtered_rmse": np.sqrt((filt_err ** 2).mean()).round(3),
    }
```

```python
# For taking average between three trials of 12 inch bed runs
stats_12_1 = summarize_file(df_live_12_1, 12)
stats_12_2 = summarize_file(df_live_12_2, 12)
stats_12_3 = summarize_file(df_live_12_3, 12)

# Average raw means

average_raw_means_12 = (stats_12_1["raw_mean"] + stats_12_2["raw_mean"] + stats_12_3["raw_mean"])/3
print("Average Raw Mean: " + str(average_raw_means_12.round(2)) + " inches")

# Average of filtered means

average_filtered_means_12 = (stats_12_1["filtered_mean"] + stats_12_2["filtered_mean"] + stats_12_3["filtered_mean"])/3
print("Average Filtered Mean: " + str(average_filtered_means_12.round(2)) + " inches")

# Average of raw biases

average_raw_biases_12 = (stats_12_1["raw_bias"] + stats_12_2["raw_bias"] + stats_12_3["raw_bias"])/3
print("Average Raw Bias: " + str(average_raw_biases_12.round(2)) + " inches")

# Average of filtered biases

average_filtered_biases_12 = (stats_12_1["filtered_bias"] + stats_12_2["filtered_bias"] + stats_12_3["filtered_bias"])/3
print("Average Filtered Bias: " + str(average_filtered_biases_12.round(2)) + " inches")

# Average of raw MAEs

average_raw_mae_12 = (stats_12_1["raw_mae"] + stats_12_2["raw_mae"] + stats_12_3["raw_mae"])/3
print("Average Raw MAE: " + str(average_raw_mae_12.round(2)) + " inches")

# Average of filtered MAEs

average_filtered_mae_12 = (stats_12_1["filtered_mae"] + stats_12_2["filtered_mae"] + stats_12_3["filtered_mae"])/3
print("Average Filtered MAE: " + str(average_filtered_mae_12.round(2)) + " inches")

# Average of raw RMSEs

average_raw_rmse_12 = (stats_12_1["raw_rmse"] + stats_12_2["raw_rmse"] + stats_12_3["raw_rmse"])/3
print("Average Raw RMSE: " + str(average_raw_rmse_12.round(2)) + " inches")

# Average of filtered RMSEs

average_filtered_rmse_12 = (stats_12_1["filtered_rmse"] + stats_12_2["filtered_rmse"] + stats_12_3["filtered_rmse"])/3
print("Average Filtered RMSE: " + str(average_filtered_rmse_12.round(2)) + " inches")

# Average of raw STD

average_raw_std_12 = (stats_12_1["raw_std"] + stats_12_2["raw_std"] + stats_12_3["raw_std"])/3
print("Average Raw STD: " + str(average_raw_std_12.round(2)) + " inches")

# Average of filtered STD

average_filtered_std_12 = (stats_12_1["filtered_std"] + stats_12_2["filtered_std"] + stats_12_3["filtered_std"])/3
print("Average Filtered STD: " + str(average_filtered_std_12.round(2)) + " inches")

# Average of samples

average_samples_12 = (stats_12_1["samples"] + stats_12_2["samples"] + stats_12_3["samples"])/3
print("Average Samples: " + str(average_samples_12) + " samples")

# Average of invalid samples

average_invalid_samples_12 = (stats_12_1["invalid_samples"] + stats_12_2["invalid_samples"] + stats_12_3["invalid_samples"])/3
print("Average Invalid Samples: " + str(average_invalid_samples_12) + " invalid samples")

# Average of invalid percentages

average_invalid_pct_12 = (stats_12_1["invalid_pct"] + stats_12_2["invalid_pct"] + stats_12_3["invalid_pct"])/3
print("Average Invalid %: " + str(average_invalid_pct_12) + " %")

```
